```python
import numpy as np
```

```python
def harmonic_mean_1(counts):
    total = 0
    count = 0
    for value in counts:
        total = total + 1/value
        count = count + 1
    return count/total
```

```python
def harmonic_mean_2(counts):
    return len(counts)/np.sum(1/counts)
```

```python
counts = np.array([1, 2, 3, 4])
assert abs(harmonic_mean_1(counts) - harmonic_mean_2(counts)) < 1e-9, harmonic_mean_1(counts)
print("Part B ok    harmonic mean =", round(harmonic_mean_1(counts), 1))
```
